## How to Test Your Work with Otter Grader

In most assignments, you will see special cells with code, `grader.check(...)`. These lines run embedded tests to to verify your solution against the expected output.

- **Running Tests**: After writing your code in a solution cell, **run the cell** (Shift+Enter). Then, run the test cells that follow it.
- **Feedback**: 
    - **Visible Tests**: If your code is correct, you will see `Test Passed`. If it's wrong, you'll see a descriptive error message.
    - **Hidden Tests**: Some tests are "hidden" and only run once you submit to Gradescope. You won't see their results locally.
- **Check All**: You can run `otter.Notebook().check_all()` if available, but usually, running individual test cells as you go is the best practice.
